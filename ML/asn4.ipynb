```python
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
```

```python
def func(x):
    return (x+3)**2

def derivative(x):
    return 2*(x+3)
```

```python
gd = []
def gradient_descent(starting_x,learning_rate,num_iterations):
    x = starting_x
    gd.append(x)
    for i in range(num_iterations):
        grad = derivative(x)
        x = x - learning_rate * grad
        gd.append(x)
        print(f"Iteration {i+1}:x = {x} f(x) = {func(x)}")
    return x;
```

```python
starting_x = 2;
learning_rate = 0.1
num_iterations = 50

min_x = gradient_descent(starting_x,learning_rate,num_iterations)
print(f"Local minimum occurs at x = {min_x}")
```

```python
x_values = np.linspace(-6, 2, 100)
y_values = func(x_values)

plt.plot(x_values, y_values)

plt.xlabel("x")
plt.ylabel("y")

plt.title("Graph of y = (x + 3)^2")

plt.grid()

plt.show()
```
